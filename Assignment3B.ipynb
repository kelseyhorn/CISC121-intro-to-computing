```python
def condenseData(dict1):
    """
    This function takes in the raw dictionary from the data file and creates a new genre key to store the genre in and
    deletes the existing genre keys into a new dictionary
    Parameters: data - a dictionary 
    Returns: dict2 - the new condensed dictionary
    """
    for row in range(len(dict1)):
        line = dict1[row]
        for key, val in line.items():
            if val == 1:
                dict1[row].update({'Genre' : key})

    info = ['title', 'rating', 'duration', 'year', 'Genre']    
    keys = dict1[0].keys()
    dict2 = dict1.copy()

    for row in range(len(dict1)):
        dict2[row] = {k:v for k,v in dict1[row].items() if v != 0 and v!= 1}
        
    return dict2
```

```python
def topGenre (watchedList, dict2):
    #new function
    """This function takes in the recently watched list and returns the
    top genre in the list"""
    genre = []
    for item in watchedList:
        for row in range(len(dict2)): 
            line = dict2[row]
            if line['title'] == item:
                genre.append(line['Genre'])
                
    return max(set(genre), key=genre.count)
```

```python
def getReccomendations (genre, watchedList, dict2):
    """This function returns the 5 most popular movies in the specified genre
    that are not on the recently watched list"""
    
    Reccomend = []
    counter = 0
    sortdict = sorted(dict2, key = lambda i: i['rating'], reverse = True) 

    for row in range(len(sortdict)):
        line = sortdict[row]
        if line['Genre'] == genre and line['title'] not in watchedList:
            Reccomend.append(line['title'])

    return Reccomend[0:5]

#get most popular movie, check against list, add to list until 5 are there
```

```python
def searchMovie(keyword, dict2):
    """This function searches the database for movies with similar titles to 
    the search parameters and returns these movies as a list"""

    title_array = []
    keyword_split = keyword.split()
    keyword_lower = [keyword.lower() for keyword in keyword_split]
    for row in range(len(dict2)): 
        line = dict2[row]
        title_array.append(line['title'])

    records = []
    for i in range(len(title_array)):
        for title in title_array:
            counter = 0
            for word in title.split():
                if word.lower() in keyword_lower:
                    counter += 1
            records.append({'Title': title, 'amount' : counter})

    sort_records = sorted(records, key = lambda i: i['amount'], reverse = True) 

    topList = []
    for row in range(len(sort_records)):
        line = sort_records[row]
        topList.append(line['Title'])

    return topList[0]
```

```python
def get10(dict2):
    top10 = []
    sortdict = sorted(dict2, key = lambda i: i['rating'], reverse = True) 
    
    for row in range(len(sortdict)):
        line = sortdict[row]
        top10.append(line['title'])
            
    return top10[0:10]
```

```python
def getShortest(dict2):
    sortdict = sorted(dict2, key = lambda i: i['duration'])

    
    return sortdict[0]['title'], sortdict[0]['duration']
```

```python
def getLongest(dict2):
    sortdict = sorted(dict2, key = lambda i: i['duration']) 
    
    return sortdict[-1]['title'], sortdict[-1]['duration']
```

```python
def getAverageLength(dict2):
    length = 0
    for row in dict2:
        length = length + row['duration']
    
    return round(length/(len(dict2)), 2)
```

```python
def markWatched (movieTitle, watchedList):
    """This function adds the movie title for watched movies into a watched list
    Parameters: title of the movie – a string
    username – a string
    watched list = a list of movies as strings
    Returns: None"""

    watchedList.append(movieTitle)
    return watchedList
```

```python
def saveWatched (watchedList, username):
    filename = 'watched' + '_' + username + '.txt'
    with open(filename, "w") as f:
        for item in watchedList:
            f.write("%s\n" % item)
    return None
```
